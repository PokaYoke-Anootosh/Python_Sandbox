```python
# QUESTION 1

# Importing the Libraries 
import openpyxl
import pandas as pd
from urllib.parse import urlparse, parse_qs

# Importing data from my Github Datafactory repo

# Cleaning the URL -
Raw_data_URL = "https://view.officeapps.live.com/op/view.aspx?src=https%3A%2F%2Fraw.githubusercontent.com%2FPokaYoke-Anootosh%2FPoka_Yoke_Data_Factory%2Frefs%2Fheads%2Fmain%2FFinance_Datasets%2FPokaYoke%2520Agencies%2520Global%2520Aging%2520raw%2520data.xlsx&wdOrigin=BROWSELINK"
Parsed_URL = urlparse(Raw_data_URL)
Parsed_URL_Query_String = Parsed_URL.query
Query_Dict = parse_qs(Parsed_URL_Query_String)
Cleaned_URL = Query_Dict["src"][0]
print(Cleaned_URL)

# Importing the Cleaned URL into a dataframe -
df_ar =  pd.read_excel(Cleaned_URL)
df_ar.head(6)
```

```python
# QUESTION 2 


# function to check for secondary address when a customer does not 
# have primary address:
def Pick_Sec_if_No_Prim(df,PrimColname,SecColname,Header="Header1"):
    df_dict = dict(df)
    Prim = df_dict[PrimColname]
    Sec = df_dict[SecColname]
    r = 0
    Consolidated_List = []

    for r in range(len(Prim)):
        if Prim[r] == None or Prim[r] == '':
            Curr = Sec[r]
        elif Sec[r] == None or Sec[r] == '':
            Curr = Prim[r]
        else:
            Curr = None

        if Header == None:
            Consolidated_List.append({
                "Header1" : Curr})
        else:
            Consolidated_List.append({
                Header : Curr})
    
    Consolidated_List_DF = pd.DataFrame(Consolidated_List)

    return Consolidated_List_DF

Address_List = Pick_Sec_if_No_Prim(
    df_ar,"Bill-to Address",
    "Ship-to Address")

Address_List.head()


    
```
